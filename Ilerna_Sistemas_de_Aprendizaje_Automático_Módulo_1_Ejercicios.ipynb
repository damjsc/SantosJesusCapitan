## Fases Entrenamiento Automático


### Importaciones

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
np.random.seed(42) # Para reproducibilidad, fijamos la semilla aleatoria
```

```python
# 1. Cargar el dataset
# load_iris() devuelve un objeto con atributos:
# - data: las características (features) de cada muestra
# - target: la etiqueta de cada muestra
# - feature_names: nombres de las características
iris = load_iris()
X = iris.data # Matriz de características, forma (150, 4)
y = iris.target # Vector de etiquetas, forma (150,)
print("Dimensiones de X:", X.shape)
print("Dimensiones de y:", y.shape)
```

```python
# 2. Mezclar y dividir los datos
# Para aprendizaje supervisado, necesitamos separar datos en:
# - Entrenamiento (train)
# - Prueba (test)
test_ratio = 0.2 # 20% de los datos para test, 80% para entrenamiento
num_samples = X.shape[0]
num_test = int(num_samples * test_ratio)
num_train = num_samples - num_test
indices = np.arange(num_samples) # Mezclamos los índices aleatoriamente
np.random.shuffle(indices)
train_indices = indices[:num_train] # Dividimos los datos
test_indices = indices[num_train:]
X_train = X[train_indices]
y_train = y[train_indices]
X_test = X[test_indices]
y_test = y[test_indices]
print("Número de muestras de entrenamiento:", X_train.shape[0])
print("Número de muestras de prueba:", X_test.shape[0])
```

```python

# 3. Crear y entrenar el modelo
model = LogisticRegression() # Usaremos regresión logística, un modelo supervisado clásico
model.fit(X_train, y_train) # Entrenamiento
```

```python
# 4. Hacer predicciones
y_pred = model.predict(X_test) # Predicciones sobre el set de prueba
```

```python
# 5. Analizamos predicciones
for i in range(len(y_test)):
  print(f"Muestra {i}: Real={y_test[i]}, Predicha={y_pred[i]}")
```

 ## Reglas de asociación con Apriori

```python
# Es una librería aparte de scikit-learn!
# Para instalarla: pip install mlxtend
from mlxtend.frequent_patterns import apriori, association_rules
import pandas as pd
# Creamos un conjunto de transacciones simplificado
dataset = [
['pan', 'leche', 'mantequilla'],
['pan', 'leche'],
['pan', 'mantequilla'],
['leche', 'galletas'],
['pan', 'galletas', 'mantequilla']
]
# Convertimos las transacciones en una matriz binaria de presencia/ausencia
items = sorted({item for transaction in dataset for item in transaction})
df = pd.DataFrame([[1 if item in t else 0 for item in items] for t in dataset], columns=items)
# Aplicamos Apriori con un soporte mínimo
frequent_itemsets = apriori(df, min_support=0.4, use_colnames=True)
# Generamos reglas de asociación con confianza y lift
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5)
print("Reglas generadas:")
print(rules[['antecedents', 'consequents', 'support', 'confidence','lift']])
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.utils import to_categorical
```

```python
# Cargar y preparar los datos
iris = load_iris()
X = iris.data
y = iris.target
# Escalado de las variables numéricas
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
# Codificación one-hot de las etiquetas
y_encoded = to_categorical(y)
```

```python
# Entrenamiento del modelo
history = model.fit(
X_train,
y_train,
epochs=100,
batch_size=8,
verbose=0,
validation_split=0.1)
```

```python
# Evaluación del modelo
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f"Accuracy del modelo en el conjunto de prueba: {accuracy:.2f}")
# Predicción sobre nuevas observaciones
predicciones = model.predict(X_test[:5])
print("\nEjemplo de predicciones (probabilidades):\n", predicciones)
```

## 7.3.2 Ejemplo de red convolucional

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
```

```python
# Cargar y preparar los datos
(X_train, y_train), (X_test, y_test) = mnist.load_data()
# Normalizar valores de píxeles (0–255) al rango [0,1]
X_train = X_train.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0
```

```python
# Redimensionar para incluir el canal (1 para imágenes en escala de grises)
X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)
# Codificar las etiquetas como vectores one-hot
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

```python
# Definir la arquitectura de la CNN
model = models.Sequential([
layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
layers.MaxPooling2D((2,2)),
layers.Conv2D(64, (3,3), activation='relu'),
layers.MaxPooling2D((2,2)),
layers.Flatten(),
layers.Dense(64, activation='relu'),
layers.Dense(10, activation='softmax')
])
```

```python
# Compilar el modelo
model.compile(optimizer='adam',
loss='categorical_crossentropy',
metrics=['accuracy'])
# Entrenar el modelo
history = model.fit(X_train, y_train, epochs=5, batch_size=64, validation_split=0.1, verbose=1)
# Evaluar el rendimiento
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"\nExactitud en el conjunto de prueba: {test_acc:.3f}")
```

## 7.4.2 Ejemplo de una RNN

```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout
```

```python
# Cargar el dataset IMDB
num_palabras = 1000 # Se considerarán las 1000 palabras más frecuentes
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=num_palabras)
```

```python
print("Tamaño del conjunto de entrenamiento:", len(X_train))
print("Ejemplo de secuencia codificada:\n", X_train[0][:20])
# Preprocesamiento: normalización de la longitud de las secuencias
max_longitud = 100 # todas las reseñas tendrán la misma longitud
X_train = pad_sequences(X_train, maxlen=max_longitud, padding='post', truncating='post')
X_test = pad_sequences(X_test, maxlen=max_longitud, padding='post', truncating='post')
```

```python
# Construcción del modelo LSTM
model = Sequential([
Embedding(input_dim=num_palabras, output_dim=64, input_length=max_longitud),
LSTM(128, return_sequences=False),
Dropout(0.2),
Dense(1, activation='sigmoid') # salida binaria: positivo/negativo
])
```

```python
# Compilación
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
# Entrenamiento
history = model.fit(X_train, y_train,
epochs=4,
batch_size=128,
validation_split=0.2,
verbose=1)
# Evaluación del modelo
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f"\nExactitud del modelo en el conjunto de prueba: {accuracy:.3f}")
```
